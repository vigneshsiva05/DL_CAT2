```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import calendar
import tensorflow as tf

from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name, sep=";")
df.columns = (
    df.columns.astype(str)
    .str.replace('"', '', regex=False)
    .str.strip()
)

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.head())
```

```python

DISTRICT_NAME = "Coimbatore"
STATE_NAME = "Tamil Nadu"

data = df[
    (df["district"].astype(str).str.strip().str.lower()
     == DISTRICT_NAME.lower()) &
    (df["state"].astype(str).str.strip().str.lower()
     == STATE_NAME.lower())
].copy()

if data.empty:
    print("Available states:", sorted(df["state"].dropna().unique()))
    print("Available districts:", sorted(df["district"].dropna().unique()))
    raise ValueError(
        "District/state not found. Check the printed names and update Cell 2."
    )

records = []

for _, row in data.iterrows():
    month = int(row["month"])

    for day in range(1, calendar.monthrange(2022, month)[1] + 1):
        if day % 100 in (11, 12, 13):
            suffix = "th"
        elif day % 10 == 1:
            suffix = "st"
        elif day % 10 == 2:
            suffix = "nd"
        elif day % 10 == 3:
            suffix = "rd"
        else:
            suffix = "th"

        column = f"{day}{suffix}"

        if column in df.columns:
            value = pd.to_numeric(row[column], errors="coerce")

            if pd.notna(value) and value < 0:
                value = np.nan

            records.append({
                "Date": pd.Timestamp(2022, month, day),
                "Rainfall": value
            })

ts = pd.DataFrame(records)
ts = ts.groupby("Date", as_index=False)["Rainfall"].mean()
ts = ts.set_index("Date").sort_index()

# Fill only missing values between known observations
ts["Rainfall"] = ts["Rainfall"].interpolate(
    method="time", limit_area="inside"
)
ts = ts.dropna()
ts = ts[ts["Rainfall"] >= 0]

print("District:", DISTRICT_NAME)
print("State:", STATE_NAME)
print("Daily records:", len(ts))
print("Date range:", ts.index.min(), "to", ts.index.max())
print("Average rainfall:", round(ts["Rainfall"].mean(), 2), "mm")

display(ts.head())
```

```python

rainfall = ts["Rainfall"].values.astype(float)

if len(rainfall) < 100:
    raise ValueError("Not enough daily records for this setup.")

WINDOW = 14
split_idx = int(len(rainfall) * 0.80)

# Log transformation reduces the effect of extreme rainfall values
rain_log = np.log1p(rainfall)

scaler = MinMaxScaler()
scaler.fit(rain_log[:split_idx].reshape(-1, 1))
scaled = scaler.transform(rain_log.reshape(-1, 1)).flatten()

X, y, target_indices = [], [], []

for i in range(WINDOW, len(scaled)):
    X.append(scaled[i-WINDOW:i])
    y.append(scaled[i])
    target_indices.append(i)

X = np.array(X).reshape(-1, WINDOW, 1)
y = np.array(y)
target_indices = np.array(target_indices)

train_mask = target_indices < split_idx
test_mask = target_indices >= split_idx

X_train_all = X[train_mask]
y_train_all = y[train_mask]
X_test = X[test_mask]
y_test = y[test_mask]

print("Training samples:", len(X_train_all))
print("Testing samples:", len(X_test))
print("Input shape:", X_train_all.shape)
```

```python

from tensorflow.keras.callbacks import Callback

# Keep the latest 15% of training data for validation
val_split = int(len(X_train_all) * 0.85)

X_fit = X_train_all[:val_split]
y_fit = y_train_all[:val_split]

X_val = X_train_all[val_split:]
y_val = y_train_all[val_split:]

class RainfallMetrics(Callback):
    def on_epoch_end(self, epoch, logs=None):
        val_pred_scaled = self.model.predict(X_val, verbose=0)

        actual_log = scaler.inverse_transform(
            y_val.reshape(-1, 1)
        ).flatten()

        predicted_log = scaler.inverse_transform(
            val_pred_scaled
        ).flatten()

        actual_mm = np.expm1(actual_log)
        predicted_mm = np.maximum(np.expm1(predicted_log), 0)

        mae = mean_absolute_error(actual_mm, predicted_mm)
        rmse = np.sqrt(mean_squared_error(actual_mm, predicted_mm))

        accuracy = np.mean(
            np.abs(actual_mm - predicted_mm) <= 5
        ) * 100

        print(
            f"\nEpoch {epoch + 1:02d} | "
            f"Accuracy (±5 mm): {accuracy:.2f}% | "
            f"MAE: {mae:.3f} mm | "
            f"RMSE: {rmse:.3f} mm"
        )

tf.keras.utils.set_random_seed(42)

model = Sequential([
    Input(shape=(WINDOW, 1)),
    LSTM(32, return_sequences=True),
    Dropout(0.15),
    LSTM(16),
    Dropout(0.10),
    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse"
)

model.summary()

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_fit,
    y_fit,
    epochs=100,
    batch_size=16,
    validation_data=(X_val, y_val),
    shuffle=False,
    callbacks=[RainfallMetrics(), early_stop],
    verbose=1
)
```

```python

pred_scaled = model.predict(X_test, verbose=0)

actual_log = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).flatten()

predicted_log = scaler.inverse_transform(
    pred_scaled
).flatten()

actual = np.maximum(np.expm1(actual_log), 0)
predicted = np.maximum(np.expm1(predicted_log), 0)

mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))

# Custom tolerance accuracy: prediction within ±5 mm
accuracy = np.mean(np.abs(actual - predicted) <= 5) * 100

print("\n========== FINAL TEST RESULTS ==========")
print(f"Test Accuracy (within ±5 mm): {accuracy:.2f}%")
print(f"Mean Absolute Error (MAE):     {mae:.3f} mm")
print(f"Root Mean Squared Error:       {rmse:.3f} mm")

results = pd.DataFrame({
    "Date": ts.index[target_indices[test_mask]],
    "Actual Rainfall (mm)": actual,
    "Predicted Rainfall (mm)": predicted,
    "Absolute Error (mm)": np.abs(actual - predicted)
})

display(results.head(20))
```

```python

plt.figure(figsize=(10, 4))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(results["Date"], results["Actual Rainfall (mm)"],
         label="Actual Rainfall")
plt.plot(results["Date"], results["Predicted Rainfall (mm)"],
         label="Predicted Rainfall")
plt.title("Actual vs Predicted Rainfall — Test Period")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
